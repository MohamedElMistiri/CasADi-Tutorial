# Solution 2 — Functions and derivatives


```python
import casadi as ca
import numpy as np
x=ca.SX.sym('x',2); value=(1-x[0])**2+100*(x[1]-x[0]**2)**2; H,g=ca.hessian(value,x)
F=ca.Function('rosenbrock',[x],[value,g,H]); out=F([1,1])
np.testing.assert_allclose(out[1].full(),0,atol=1e-12)
eps=1e-6; x0=np.array([0.8,1.1]); fd=np.array([(float(F(x0+eps*np.eye(2)[i])[0])-float(F(x0-eps*np.eye(2)[i])[0]))/(2*eps) for i in range(2)])
np.testing.assert_allclose(fd,F(x0)[1].full().ravel(),rtol=1e-5)
assert F.map(4)(ca.DM.zeros(2,4))[0].shape==(1,4)
```
