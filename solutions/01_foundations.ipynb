# Solution 1 — Symbolic matrices and sparsity


```python
import casadi as ca
from casadi_tutorial.plotting import plot_sparsity
x=ca.SX.sym('x',5); f=ca.vertcat(x[0]*x[1],ca.sin(x[2])+x[4]**2); J=ca.jacobian(f,x)
print(J.shape,J.nnz()); assert J.shape==(2,5) and J.nnz()==4
assert all(row != 3 for row in J.sparsity().get_triplet()[1])
plot_sparsity(J);
```
